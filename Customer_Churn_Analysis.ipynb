```python

import pandas as pd

df = pd.read_csv('/content/Telco-Customer-Churn.csv')

print("Dataset loaded successfully!")
print("Dataset Shape:", df.shape)

df.head()
```

```python

# Display all column names
print("Column Names:")
print(df.columns.tolist())

# Check data types and missing values
print("\nDataset Information:")
df.info()
```

```python

print("First 5 Records:")
display(df.head())

print("Last 5 Records:")
display(df.tail())
```

```python

print("Number of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])
```

```python

# Check missing values in each column
print(df.isnull().sum())
```

```python

# Count duplicate rows
print("Duplicate rows:", df.duplicated().sum())
```

```python

# Check data types
print(df.dtypes)

# Convert TotalCharges into numeric format
df['TotalCharges'] = pd.to_numeric(
    df['TotalCharges'], errors='coerce'
)

# Check missing values again
print("\nMissing values after conversion:")
print(df.isnull().sum())
```

```python

# Remove rows with missing TotalCharges
df = df.dropna(subset=['TotalCharges'])

print("Cleaned Dataset Shape:", df.shape)
print("Remaining Missing Values:", df.isnull().sum().sum())
```

```python

# Count customers based on churn status
print(df['Churn'].value_counts())
```

```python

# Calculate churn percentage
churn_percentage = df['Churn'].value_counts(normalize=True) * 100

print("Customer Churn Percentage:")
print(churn_percentage)

print("\nOverall Churn Rate:",
      round(churn_percentage['Yes'], 2), "%")
```

```python

import matplotlib.pyplot as plt

churn_counts = df['Churn'].value_counts()

plt.figure(figsize=(7,5))
plt.bar(churn_counts.index, churn_counts.values)

plt.title("Customer Churn Analysis")
plt.xlabel("Churn Status")
plt.ylabel("Number of Customers")

plt.show()
```

```python

contract_churn = pd.crosstab(
    df['Contract'],
    df['Churn'],
    normalize='index'
) * 100

print(contract_churn)
```

```python

contract_churn.plot(
    kind='bar',
    figsize=(8,5)
)

plt.title("Customer Churn by Contract Type")
plt.xlabel("Contract Type")
plt.ylabel("Percentage of Customers")
plt.xticks(rotation=0)
plt.legend(title="Churn")
plt.show()
```

```python

tenure_churn = df.groupby('Churn')['tenure'].mean()

print("Average Customer Tenure:")
print(tenure_churn)
```

```python

df.boxplot(
    column='tenure',
    by='Churn',
    figsize=(7,5)
)

plt.title("Customer Tenure vs Churn")
plt.suptitle("")
plt.xlabel("Churn Status")
plt.ylabel("Tenure (Months)")
plt.show()
```

```python

monthly_charges = df.groupby('Churn')['MonthlyCharges'].mean()

print("Average Monthly Charges:")
print(monthly_charges)
```

```python

monthly_charges.plot(
    kind='bar',
    figsize=(7,5)
)

plt.title("Average Monthly Charges by Churn")
plt.xlabel("Churn Status")
plt.ylabel("Average Monthly Charges")
plt.xticks(rotation=0)
plt.show()
```

```python

total_charges = df.groupby('Churn')['TotalCharges'].mean()

print("Average Total Charges:")
print(total_charges)
```

```python

total_charges.plot(
    kind='bar',
    figsize=(7,5)
)

plt.title("Average Total Charges by Churn")
plt.xlabel("Churn Status")
plt.ylabel("Average Total Charges")
plt.xticks(rotation=0)
plt.show()
```

```python

internet_churn = pd.crosstab(
    df['InternetService'],
    df['Churn'],
    normalize='index'
) * 100

print(internet_churn)
```

```python

internet_churn.plot(
    kind='bar',
    figsize=(8,5)
)

plt.title("Customer Churn by Internet Service")
plt.xlabel("Internet Service")
plt.ylabel("Churn Percentage")
plt.xticks(rotation=0)
plt.show()
```

```python

payment_churn = pd.crosstab(
    df['PaymentMethod'],
    df['Churn'],
    normalize='index'
) * 100

print(payment_churn)
```

```python

payment_churn.plot(
    kind='bar',
    figsize=(10,5)
)

plt.title("Customer Churn by Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Churn Percentage")
plt.xticks(rotation=30, ha='right')
plt.show()
```

```python

print("CUSTOMER CHURN ANALYSIS REPORT")
print("-" * 40)

print("Total Customers:", len(df))

print("Customers Who Left:",
      (df['Churn'] == 'Yes').sum())

print("Customers Who Stayed:",
      (df['Churn'] == 'No').sum())

print("Overall Churn Rate:",
      round((df['Churn'] == 'Yes').mean() * 100, 2), "%")

print("Average Monthly Charges:",
      round(df['MonthlyCharges'].mean(), 2))

print("Average Customer Tenure:",
      round(df['tenure'].mean(), 2), "months")
```
